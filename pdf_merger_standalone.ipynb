# PDF Merger (single cell)
Set `FOLDER`, then run the cell. Files named `A_...`, `B_...`, `C_...` merge in that order.

```python
# pip install pypdf   (run once, if needed)
%pip install pypdf
import re
from pathlib import Path
from pypdf import PdfReader, PdfWriter

FOLDER = r"C:\Users\kwind\OneDrive\Documents\pdf merge\A"   # <-- your folder
OUTPUT_NAME = "merged.pdf"                                       # saved inside that folder


def natural_key(text):
    return [int(t) if t.isdigit() else t.lower() for t in re.split(r"(\d+)", text)]


def sort_key(name):
    m = re.match(r"^([A-Za-z])_", name)
    return (m.group(1).upper() if m else "~", natural_key(name))   # A_, B_, C_ ... then the rest


folder = Path(FOLDER.strip().strip('"'))
if not folder.is_dir():
    raise FileNotFoundError(f"Folder not found: {folder}")

files = sorted(
    (p for p in folder.glob("*.pdf") if p.name.lower() != OUTPUT_NAME.lower()),
    key=lambda p: sort_key(p.name),
)
if not files:
    raise ValueError("No PDF files found in that folder.")

print("Merge order:")
for i, f in enumerate(files, 1):
    print(f"  {i}. {f.name}")

writer = PdfWriter()
for f in files:
    start = len(writer.pages)
    for page in PdfReader(str(f)).pages:
        writer.add_page(page)
    writer.add_outline_item(f.stem, start)      # bookmark per file

output = folder / OUTPUT_NAME
with open(output, "wb") as fh:
    writer.write(fh)

print(f"\nSaved: {output}  ({len(writer.pages)} pages)")
```
